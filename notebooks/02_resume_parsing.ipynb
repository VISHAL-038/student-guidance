```python
import spacy
import re
import json
import pandas as pd
from sentence_transformers import SentenceTransformer


```

```python
nlp = spacy.load("en_core_web_sm")

```

```python
SKILL_KEYWORDS = [
    "python", "java", "c++", "django", "machine learning",
    "deep learning", "nlp", "data science", "sql", "html",
    "css", "javascript", "react", "pandas", "numpy", "tensorflow"
]

```

```python
sample_resume = """
Vishal 
M.Tech Computer Science

Skills: Python, Machine Learning, Django, SQL, Deep Learning, HTML, CSS
Experience: Worked on ML projects involving classification, NLP, and recommendation systems.
Education: B.Tech (CSE), M.Tech (Pursuing)
"""
```

```python
def extract_skills(text):
    found = []
    text_lower = text.lower()
    for skill in SKILL_KEYWORDS:
        if skill in text_lower:
            found.append(skill)
    return list(set(found))

skills = extract_skills(sample_resume)
skills
```

```python
def extract_education(text):
    education_patterns = [
        r"(B\.?Tech|Bachelor|Bachelors|BSc|B\.Sc)",
        r"(M\.?Tech|Master|Masters|MSc|M\.Sc|Postgraduate)"
    ]
    
    found = []
    for pattern in education_patterns:
        match = re.findall(pattern, text, flags=re.IGNORECASE)
        found.extend(match)
    return list(set(found))

education = extract_education(sample_resume)
education
```

```python
def extract_experience(text):
    exp_pattern = r"(experience:.*)"
    match = re.findall(exp_pattern, text, re.IGNORECASE)
    return match[0] if match else "Not found"

experience = extract_experience(sample_resume)
experience
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")

resume_embedding = model.encode(sample_resume)
len(resume_embedding)
```

```python
resume_json = {
    "skills": skills,
    "education": education,
    "experience": experience,
    "embedding": resume_embedding.tolist()
}

resume_json

```
