```python
# Clean column names
students_df.columns = [c.strip().lower().replace(" ", "_") for c in students_df.columns]

# Fill missing numeric values with column mean
students_df = students_df.fillna(students_df.mean(numeric_only=True))

# Add an average score column
students_df["average_score"] = students_df[["math_score", "reading_score", "writing_score"]].mean(axis=1)

# Save the cleaned version
os.makedirs("../data/processed", exist_ok=True)
students_df.to_csv("../data/processed/students_performance_clean.csv", index=False)

print("✅ Cleaned dataset saved successfully.")

```

```python
import pandas as pd
import os

def load_and_clean_student_data(raw_path="../data/raw/students_performance.csv"):
    df = pd.read_csv(raw_path)
    df.columns = [c.strip().lower().replace(" ", "_") for c in df.columns]
    df = df.fillna(df.mean(numeric_only=True))
    df["average_score"] = df[["math_score", "reading_score", "writing_score"]].mean(axis=1)
    
    os.makedirs("../data/processed", exist_ok=True)
    cleaned_path = "../data/processed/students_performance_clean.csv"
    df.to_csv(cleaned_path, index=False)
    print(f"✅ Cleaned data saved to {cleaned_path}")
    return df
```
